```python
from pyspark.sql.functions import col
```

```python
# Create a delta table for diet features
@SWSAcademyHelper.add_init
def create_security_features_table(self):
    
    spark.sql(f"USE CATALOG {SA.catalog_name}")
    spark.sql(f"USE SCHEMA {SA.schema_name}")

    # The demo creates this feature table, drops a column and reads its version history,
    # so start from a fresh table on every run.
    spark.sql(f"DROP TABLE IF EXISTS {SA.catalog_name}.{SA.schema_name}.telco_customer_features")
    # An overwrite keeps the primary key the demo adds, so recreate this table too.
    spark.sql(f"DROP TABLE IF EXISTS {SA.catalog_name}.{SA.schema_name}.security_features")

    # Load dataset with spark
    shared_volume_name = 'telco' # From Marketplace
    csv_name = 'telco-customer-churn-missing' # CSV file name
    dataset_path = f"{SA.paths.datasets.telco}/{shared_volume_name}/{csv_name}.csv" # Full path

    df = spark.read.csv(dataset_path, header="true", inferSchema="true", multiLine="true", escape='"')

    # drop the taget column
    df = df.select("customerID", "OnlineSecurity", "OnlineBackup", "DeviceProtection")

    table_name = f"{SA.catalog_name}.{SA.schema_name}.security_features"
    df.write.format("delta").mode("overwrite").saveAsTable(table_name)
```
