```python
# Log this notebook's runs to a named experiment in your home folder. A notebook's built-in experiment cannot be
# restored once deleted (for example by a cleanup), which would stop the notebook from logging runs; a named
# experiment is restored automatically if a cleanup removed it.
import mlflow
from mlflow.tracking import MlflowClient
_nb_name = dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get().rsplit("/", 1)[-1]
_exp_name = f"/Users/{spark.sql('SELECT current_user()').first()[0]}/SWSAcademy - {_nb_name}"
_exp = MlflowClient().get_experiment_by_name(_exp_name)
if _exp is not None and _exp.lifecycle_stage == "deleted":
    MlflowClient().restore_experiment(_exp.experiment_id)
mlflow.set_experiment(_exp_name)
print("MLflow experiment:", _exp_name)
```
